```python
#Write a program that finds the factorial of a given number.
number = int(input("Enter a number: "))
factorial = 1
for i in range(1, number + 1):
    factorial = factorial * i
print("Factorial:", factorial)
```

```python
# Write a program to perform the given operations on a list:i. Addition ii. Insertion iii. Slicing
numbers = [10, 20, 30]
print("Original list:", numbers)
numbers = numbers + [40]
print("After addition:", numbers)
numbers.insert(1, 15)
print("After insertion:", numbers)
print("Slicing:", numbers[1:4])
```

```python
#Write a program that generates electricity bill based on the meter readings.
name = input("Enter customer name: ")
meter = input("Enter meter number: ")
old = int(input("Enter previous reading: "))
new = int(input("Enter current reading: "))

units = new - old

if units <= 50:
    bill = units * 2.30
elif units <= 200:
    bill = (50 * 2.30) + (units - 50) * 4.00
elif units <= 400:
    bill = (50 * 2.30) + (150 * 4.00) + (units - 200) * 5.00
else:
    bill = (50 * 2.30) + (150 * 4.00) + (200 * 5.00) + (units - 400) * 5.40

print("Customer name:", name)
print("Meter number:", meter)
print("Units used:", units)
print("Bill amount: Rs.", bill)
```

```python
#Write a program to perform any 5 built-in functions by taking any list.
numbers = [10, 20, 30, 40, 50]
print("Length:", len(numbers))
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))
```

```python
#Write a program that performs the following list methods:Append,Insert,Extend,Count,Sort
numbers = [3, 1, 2]
numbers.append(4)
print("Append:", numbers)
numbers.insert(1, 5)
print("Insert:", numbers)
numbers.extend([6, 7])
print("Extend:", numbers)
print("Count of 2:", numbers.count(2))
numbers.sort()
print("Sort:", numbers)
```

```python
#Write a program to find the length of the string without using any library functions.
text = input("Enter a string: ")
count = 0
for ch in text:
    count = count + 1
print("Length of string:", count)
```

```python
#Write a program to check if the substring is present in a given string or not.
text = input("Enter a string: ")
sub = input("Enter substring: ")
if sub in text:
    print("Substring is present")
else:
    print("Substring is not present")
```

```python
#Write a program to count number of vowels in a given string.
text = input("Enter a string: ")
count = 0
for ch in text:
    if ch in 'aeiouAEIOU':
        count = count + 1
print("Number of vowels:", count)
```

```python
#Write a program to check if two strings are ANAGRAMS or not.
a = input("Enter first string: ")
b = input("Enter second string: ")
a = a.lower()
b = b.lower()
if sorted(a) == sorted(b):
    print("Anagrams")
else:
    print("Not anagrams")
```

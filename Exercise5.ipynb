```python
#Write a python program to find number of duplicates present in the list
numbers = [1, 2, 2, 3, 3, 3, 4]
duplicates = 0
checked = []
for num in numbers:
    if numbers.count(num) > 1 and num not in checked:
        duplicates = duplicates + 1
        checked.append(num)
print("Number of duplicate elements:", duplicates)
```

```python
#Write a program to check if a given key exists in a dictionary or not.
student = {"name": "Ravi", "age": 20, "course": "Python"}
key = input("Enter a key: ")

if key in student:
    print("Key exists")
else:
    print("Key does not exist")
```

```python
#Write a program to sum all the items in a given dictionary.
marks = {"Maths": 80, "Science": 90, "English": 70}
total = 0

for value in marks.values():
    total = total + value

print("Sum of dictionary values:", total)
```

```python
# Write a python program that creates a dictionary using dictionary comprehension.
squares = { }
for i in range(1, 6):
    squares[i] = i * i
print(squares)
```

```python
#Write a python program that prints frequency of each element in a list in dictionary format.
numbers = [1, 2, 2, 3, 1, 2]
frequency = { }
for num in numbers:
    if num in frequency:
        frequency[num] = frequency[num] + 1
    else:
        frequency[num] = 1
print(frequency)
```

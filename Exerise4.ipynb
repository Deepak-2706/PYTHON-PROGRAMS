```python
# Write a program to check given number is palindrome or not using functions.
def palindrome(number):
    reverse = 0
    temp = number
    while temp > 0:
        digit = temp % 10
        reverse = reverse * 10 + digit
        temp = temp // 10

    if number == reverse:
        return True
    else:
        return False
n = int(input("Enter a number: "))

if palindrome(n):
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
#Write a program to return the reverse of the string using functions.
def reverse_string(text):
    return text[::-1]
word = input("Enter a string: ")
print("Reverse string:", reverse_string(word))
```

```python
#Write a program to define a function with multiple return values.
def calculate(a, b):
    sum = a + b
    difference = a - b
    product = a * b
    return sum, difference, product
x, y, z = calculate(10, 5)
print("Sum:", x)
print("Difference:", y)
print("Product:", z)
```

```python
#. Write a program that covers read mode, write mode, append mode in file handling.
file = open("sample.txt", "w")
file.write("Hello Python\n")
file.close()
file = open("sample.txt", "a")
file.write("Welcome to file handling")
file.close()
file = open("sample.txt", "r")
print(file.read())
file.close()
```

```python
#Write a Python program to print each line of a file in reverse order.
file = open("sample.txt", "r")
lines = file.readlines()
file.close()
for line in lines[::-1]:
    print(line.strip())
```

```python
#Write a Python program to compute the number of characters, words and lines in a file.
file = open("sample.txt", "r")
text = file.read()
file.close()
characters = len(text)
words = len(text.split())
lines = len(text.splitlines())
print("Characters:", characters)
print("Words:", words)
print("Lines:", lines)
```
